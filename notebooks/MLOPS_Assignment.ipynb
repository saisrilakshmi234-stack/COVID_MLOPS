## 16.Data Preprocessing

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

numerical_features = ["Confirmed last week"]
categorical_features = ["WHO Region"]

preprocessor = ColumnTransformer([
    ("num", SimpleImputer(strategy="median"), numerical_features),
    ("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]), categorical_features)
])

print("Data preprocessing setup completed successfully.")
```

```python
# Preprocess training and testing data

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Data preprocessing completed successfully.")
```

In this step, we prepare the input features for machine learning model.
- Libraries used: 
- from sklearn.compose import ColumnTransformer
'ColumnTransfer" allows us to apply different preprocessing operations to different types of columns. Here, it is used to apply one-hot encoding only to the categorical features while keeping the numerical features unchanged.
- from sklearn.preprocessing import OneHotEncoder
"OneHotEncoder" converts categorical text values into numerical features.

The numerical feature "Confirmed last week" is handled using median imputation, while the categorical feature "WHO Region" is handled using most-frequent imputation and One-Hot Encoding.

## 20. Model evaluation metrics

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)
models = {
    "Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge(alpha=1.0),
    "Lasso Regression": Lasso(alpha=1.0, max_iter=10000),
    "Random Forest Regression": RandomForestRegressor(
        n_estimators=300,
        max_depth=10,
        min_samples_split=5,
        min_samples_leaf=2,
        random_state=42
    )
}
model_results = []
for name, model in models.items():
    model.fit(X_train_processed, y_train)
    y_pred = model.predict(X_test_processed)
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)
    model_results.append({
        "Model": name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2 Score": r2
    })
results_df = pd.DataFrame(model_results)
results_df = results_df.sort_values(
    by="R2 Score",
    ascending=False
).reset_index(drop=True)
display(results_df.round(4))
best_model_name = results_df.loc[0, "Model"]
best_r2 = results_df.loc[0, "R2 Score"]
print("Best Model:", best_model_name)
print("Best R2 Score:", round(best_r2, 4))
```

The trained models are evaluated using performance metrics are mse,mae,rmse,r2 score.
- Libraries used
- mean_absolute_error
It calculates the average absolute difference between actual and predicted errors.
-  mean_squared_error
It calculates the average squared difference between actual and predicted values.
- r2_score
it measures how well the model expalins the variation in the target variable.
- numpy
It used to calculate rmse and mse

## 21. Cross valiadtion

```python
from sklearn.model_selection import cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor
import pandas as pd

cv_models = {
    "Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge(alpha=1.0),
    "Lasso Regression": Lasso(alpha=1.0, max_iter=10000),
    "Random Forest Regression": RandomForestRegressor(
        n_estimators=300,
        max_depth=10,
        min_samples_split=5,
        min_samples_leaf=2,
        random_state=42
    )
}

cv_results = []

for name, model in cv_models.items():

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    scores = cross_val_score(
        pipeline,
        X,
        y,
        cv=5,
        scoring="r2"
    )

    cv_results.append({
        "Model": name,
        "Fold 1": scores[0],
        "Fold 2": scores[1],
        "Fold 3": scores[2],
        "Fold 4": scores[3],
        "Fold 5": scores[4],
        "Mean R2 Score": scores.mean()
    })

cv_results_df = pd.DataFrame(cv_results)

display(cv_results_df.round(4))

best_cv_model = cv_results_df.loc[
    cv_results_df["Mean R2 Score"].idxmax(),
    "Model"
]

best_cv_score = cv_results_df["Mean R2 Score"].max()

print("Best Cross-Validated Model:", best_cv_model)
print("Mean R2 Score:", round(best_cv_score, 4))
```

- Libraries used
- from sklearn.model_selection import cross_val_score
It is used to perform cross-validation and calculate the R2 score for each fold.

In this process, the dataset is divided into multiple folds. The model is trained on some folds and tested on the remaining fold. This process is repeated so that every fold is used for validation.

Here, 5-Fold Cross-Validation is used. The R² score is calculated for each fold and the mean R² score is used to understand the overall model performance.

A higher mean R² score indicates better and more consistent model performance.
